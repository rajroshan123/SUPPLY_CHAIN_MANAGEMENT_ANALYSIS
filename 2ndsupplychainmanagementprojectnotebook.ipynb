```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(" ","_")
df.columns = df.columns.str.replace (r"[()]", "", regex=True)
```

```python
df["customer_lname"] = df["customer_lname"].fillna("unknown")
#  AS AFTER OBSERVING I CAME TO KNOW THAT ZIPCODE DEPENDS ON THE LATITUDE AND LONGITUDINAL VALUE SO I USED KNN FOR PREDICTING THE MISSING VALUES BASED ON THE VALUES OF LATITUDE AND LONGITUDE
from sklearn.neighbors import KNeighborsRegressor

x = df[["latitude", "longitude"]]
y = df["customer_zipcode"]

known = y.notna()
missing = y.isna()

model = KNeighborsRegressor(n_neighbors=1)
model.fit(x[known], y[known])

df.loc[missing, "customer_zipcode"] = model.predict(x[missing]).round()
```

```python
#now we shall create some columns so that we can take best insights further in sql or power bi
#1st one shipping delay days
df["shipping_delay_days"]=(df["days_for_shipping_real"]-df["days_for_shipment_scheduled"])
# 2nd profit per order item
df["profit_per_order_item"]=(df["order_item_product_price"]*df["order_item_profit_ratio"])
# 3rd order month
df["order_date_dateorders"]=pd.to_datetime(df["order_date_dateorders"],errors="coerce")
df["order_year_month"] = (df["order_date_dateorders"].dt.to_period("M").astype(str))
# 4th sales category
df["sales_category"]=df["sales_per_customer"].apply(lambda x: "low" if x<250 else "medium" if x<=500 else "high")
# delievery performance
df["delievery_performance"]=df["shipping_delay_days"].apply(lambda x:"late" if x>0 else "on time")
df[["shipping_delay_days",
    "profit_per_order_item",
    "order_year_month",
    "sales_category",
    "delievery_performance"
  ]].head(70)
```
